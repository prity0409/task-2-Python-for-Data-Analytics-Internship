```python
import pandas as pd

# Titanic dataset load karne ke liye
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

# Pehli 5 rows dekhne ke liye
df.head()
```

(Data Manipulation and Analysis)

```python
# To check the basic information and missing (null) values of the data
print("--- Data Info ---")
df.info()

# To view the summary statistics
print("\n--- Summary Statistics ---")
df.describe()
```

```python
#  TO SEE THE Columns  NAME 
df.columns
```

```python
# To check the number of rows and columns
df.shape
```

```python
# To check missing values in each column
df.isnull().sum()
```

```python
# To filter passengers who survived (Survived == 1)
df[df['Survived'] == 1].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# To plot a count of passengers who survived
sns.countplot(x='Survived', data=df)
plt.show()
```

```python
# To plot the age distribution of passengers
sns.histplot(df['Age'].dropna(), bins=30, kde=True)
plt.show()
```
